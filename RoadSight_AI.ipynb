```python
import os
from dotenv import load_dotenv

load_dotenv()

ROBOFLOW_API = os.getenv("ROBOFLOW_API_KEY")
```

```python
!pip install roboflow

from roboflow import Roboflow
rf = Roboflow(api_key=ROBOFLOW_API)
project = rf.workspace("new-workspace-kj87b").project("road-damage-detection-iicdh")
version = project.version(10)
dataset = version.download("yolov8")
```

```python
import yaml

with open(dataset.location + "/data.yaml", "r") as f:
    data_config = yaml.safe_load(f)

print(data_config)
print("Classes:", data_config["names"])
print("Number of classes:", data_config["nc"])
```

```python
!pip install ultralytics
```

```python
from ultralytics import YOLO

model = YOLO("yolov8s.pt")

results = model.train(
    data=dataset.location + "/data.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    patience=20,
    device=0,
    project="RoadSight_AI",
    name="RDD2022_YOLOv8s"
)
```
